## Preliminaries: supplied data and helpers

Use the supplied coefficients consistently. Positive means default throughout. Use plausible ranges, such as DTI between 0.10 and 0.65 and bureau score between 400 and 850. Keep interventions within these ranges, and separate model association from causal interpretation.

```python
from scipy.special import expit

MODEL = {'intercept': -2.0, 'dti_coefficient': 5.0,
         'bureau_per_100_coefficient': -0.8}

def linear_predictor(dti, bureau_score):
    return (MODEL['intercept'] + MODEL['dti_coefficient'] * np.asarray(dti)
            + MODEL['bureau_per_100_coefficient']
            * ((np.asarray(bureau_score) - 650) / 100))

def predict_default(dti, bureau_score):
    return expit(linear_predictor(dti, bureau_score))

profiles = pd.DataFrame({'profile': ['A', 'B', 'C'],
                         'dti': [0.15, 0.35, 0.55],
                         'bureau_score': [750, 650, 550]})
display(profiles)
# Examples of function use; design your own controlled comparisons below.
print('Predicted default probability at DTI=0.30, score=650:',
      round(float(predict_default(0.30, 650)), 4))
```

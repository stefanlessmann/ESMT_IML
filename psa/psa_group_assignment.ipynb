## Data of the groups and topics

```python
import pandas as pd
from IPython.display import display
import random

# Data for groups and topics
group_names = [
    "Chardonnay", "Chianti", "Grauburgunder", "Pinot Grigio", "Pinot Nero",
    "Primitivo", "Prosecco", "Riesling", "Rioja",
]
groups = [(f"G{i}", name) for i, name in enumerate(group_names, start=1)]
randomized_groups = groups.copy() # a copy that we randomize below
slots = [f"S{i}" for i in range(1, len(groups) + 1)]
topics = [
    "S1. How to handle missing values in a dataset?",
    "S2. How to deal with imbalanced datasets?",            
    "S3. How to interpret the results of a machine learning model?",
    "S4. How to evaluate the performance of a machine learning model?",
    "S5. How to avoid overfitting in a machine learning model?",
    "S6. How to choose the right machine learning algorithm for a given problem?",
    "S7. How to preprocess data for machine learning?",
    "S8. How to visualize data in Python?",
    "S9. How to work with large datasets in Python?"
]   
```

## Randomization 
We create a **random** ordering of study groups and then assign the groups in this random order to PSA slots.

```python

# *********************************************************+
# RANDOMIZE THE GROUPS
# *********************************************************+
SEED = 67
random.seed(SEED)
random.shuffle(randomized_groups) # rearrange a list randomly in place 
```

## Display the allocation

```python
assignments = pd.DataFrame(randomized_groups, columns=["Group number", "Group name"], index=slots)
assignments["Practice session augmentation topic"] = topics
display(assignments)
```
